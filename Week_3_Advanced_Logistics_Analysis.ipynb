```python
import pandas as pd

df = pd.read_csv('/content/logistics_data/Delivery_Logistics.csv')

print("Dataset Shape:", df.shape)
```

```python
df['delivery_time_hours'] = pd.to_datetime(
    df['delivery_time_hours']
).dt.nanosecond

df['expected_time_hours'] = pd.to_datetime(
    df['expected_time_hours']
).dt.nanosecond

print(df[['delivery_time_hours', 'expected_time_hours']].head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.boxplot(df['delivery_time_hours'])

plt.title('Box Plot of Delivery Time')
plt.ylabel('Delivery Time (Hours)')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,7))

corr = df[
    ['distance_km',
     'package_weight_kg',
     'delivery_time_hours',
     'expected_time_hours',
     'delivery_rating',
     'delivery_cost']
].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Logistics Variables')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    df['distance_km'],
    df['delivery_time_hours'],
    alpha=0.4
)

plt.title('Distance vs Delivery Time')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Time (Hours)')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(
    df['distance_km'],
    df['delivery_cost'],
    alpha=0.4
)

plt.title('Distance vs Delivery Cost')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Cost')

plt.show()
```

```python
weather_delay = (
    df.groupby('weather_condition')['delayed']
    .mean() * 100
)

print(weather_delay.sort_values(ascending=False))
```

```python
df['delayed_numeric'] = df['delayed'].map({
    'yes': 1,
    'no': 0
})

print(df[['delayed', 'delayed_numeric']].head())
```

```python
weather_delay = (
    df.groupby('weather_condition')['delayed_numeric']
    .mean() * 100
)

print(weather_delay.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

weather_delay_sorted = weather_delay.sort_values(ascending=False)

plt.figure(figsize=(8,5))

plt.bar(
    weather_delay_sorted.index,
    weather_delay_sorted.values
)

plt.title('Delivery Delay Rate by Weather Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Delay Rate (%)')

plt.xticks(rotation=30)
plt.show()
```

```python
partner_delay = (
    df.groupby('delivery_partner')['delayed_numeric']
    .mean() * 100
)

print(partner_delay.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

partner_delay_sorted = partner_delay.sort_values(ascending=False)

plt.figure(figsize=(10,5))

plt.bar(
    partner_delay_sorted.index,
    partner_delay_sorted.values
)

plt.title('Delivery Delay Rate by Delivery Partner')
plt.xlabel('Delivery Partner')
plt.ylabel('Delay Rate (%)')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
region_delay = (
    df.groupby('region')['delayed_numeric']
    .mean() * 100
)

print(region_delay.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

region_delay_sorted = region_delay.sort_values(ascending=False)

plt.figure(figsize=(8,5))

plt.bar(
    region_delay_sorted.index,
    region_delay_sorted.values
)

plt.title('Delivery Delay Rate by Region')
plt.xlabel('Region')
plt.ylabel('Delay Rate (%)')

plt.tight_layout()
plt.show()
```

```python
print("Overall Delay Rate:")
print(df['delayed_numeric'].mean() * 100)

print("\nAverage Delivery Time:")
print(df['delivery_time_hours'].mean())

print("\nAverage Delivery Cost:")
print(df['delivery_cost'].mean())

print("\nAverage Delivery Rating:")
print(df['delivery_rating'].mean())
```

```python
print("KEY ANALYTICAL INSIGHTS")
print("-----------------------")

print("1. Overall delay rate:", round(df['delayed_numeric'].mean() * 100, 2), "%")
print("2. Average delivery time:", round(df['delivery_time_hours'].mean(), 2), "hours")
print("3. Average delivery cost:", round(df['delivery_cost'].mean(), 2))
print("4. Average delivery rating:", round(df['delivery_rating'].mean(), 2), "/ 5")

print("\nHighest weather delay rate:")
print(weather_delay.idxmax(), "=", round(weather_delay.max(), 2), "%")

print("\nHighest partner delay rate:")
print(partner_delay.idxmax(), "=", round(partner_delay.max(), 2), "%")

print("\nHighest region delay rate:")
print(region_delay.idxmax(), "=", round(region_delay.max(), 2), "%")
```
